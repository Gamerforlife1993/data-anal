# Bar plot representing the number of different types of Universities across the state of West Bengal compared to India

# Pie charts representing the percentage distribution of different University types across West Bengal compared to India

```python
fig, axs = plt.subplots(2, 4, figsize=(14, 10))


dataframes = [SPU_C, SPR_C, INI_C, CU_C, DUG_C, DUP_C, SOU_C]
labels = ['State Public University', 'State Private University', 'Institute Of National Importance', 
          'Central University', 'Deemed University Govt', 'Deemed University Private', 'State Open University']


for i, ax in enumerate(axs.flat):
    if i < len(dataframes):  
        df = dataframes[i]
        label = labels[i]
        df['University Type'].value_counts(normalize=True).plot(kind='pie', autopct='%.2f%%', ax=ax,wedgeprops={'edgecolor':'black'})
        ax.set_xlabel(label)  
        ax.set_aspect('equal') 
    else:
        ax.axis('off')  

plt.tight_layout()
plt.show()
```

```python
import geopandas as gpd
from geopy.geocoders import Nominatim



geolocator = Nominatim(user_agent="university_geocoder")
state_counts['Coordinates'] = state_counts['State'].apply(lambda x: geolocator.geocode(x).point if geolocator.geocode(x) else None)


# Create a map
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))
ax = world[world.name == "India"].plot(color='white', edgecolor='black')

# Plotting points for each state
for idx, row in state_counts.iterrows():
    ax.text(row['Coordinates'][1], row['Coordinates'][0], s=row['State'], horizontalalignment='center', fontsize=8, color='red')

# Show plot
ax.set_title("Distribution of Universities by State")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.show()
```
